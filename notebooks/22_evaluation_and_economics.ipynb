```python
# From the slide: The numeric-claim linter
import math, re
NUM = re.compile(r"""(?P<neg>\()?\$?\s*
  (?P<num>\d{1,3}(?:,\d{3})+(?:\.\d+)?|\d+(?:\.\d+)?)
  \s*(?P<unit>%|x\b|bn\b|mm\b|m\b|million\b|k\b)?
  (?(neg)\))""", re.X | re.I)
YEAR = re.compile(r"^(19|20)\d{2}$")
SCALE = {"bn": 1e9, "mm": 1e6, "m": 1e6, "million": 1e6,
         "k": 1e3}

def numbers_in(text):
    for m in NUM.finditer(text):
        raw, unit = m["num"], (m["unit"] or "").lower()
        a, b = m.start("num"), m.end("num")
        if (YEAR.match(raw) and not unit
                or text[a-1:a].isalpha()   # Q2, FY24
                or re.match(r"-[A-Za-z]", text[b:b+2])
                or re.search(r"(item|note)\s*$",
                             text[max(0, a-6):a], re.I)):
            continue               # 10-K, 8-K, Item 7
        v = float(raw.replace(",", ""))
        v *= SCALE.get(unit, 1)
        yield -v if m["neg"] else v

def lint(text, allowed, tol=0.005):     # [] == pass
    ok = lambda v: any(math.isclose(v, a, rel_tol=tol)
                       for a in allowed)
    return [v for v in numbers_in(text) if not ok(v)]
```

```python
answers = [json.loads(l) for l in open("outputs/qa/dev.jsonl")]
text = {(r["doc_id"], r["block_id"]): r["text"] for r in sd.records("deal_a")}   # block ids repeat across documents
cells = {f["citation"]["cell"]: f["value"] for f in sd.seller_facts() if "cell" in f["citation"]}
def supports(cit, value):
    if cit["block_id"] in cells: return abs(cells[cit["block_id"]] - value) <= 0.005 * abs(value)
    return any(abs(v - value) <= 0.005 * abs(value) or abs(v - value * 100) <= 0.005 * abs(value * 100) for v in numbers_in(text.get((cit["doc_id"], cit["block_id"]), "")))
cited = supported = claims_with_support = n_claims = 0
for a in answers:
    for c in a["answer"]["claims"]:
        n_claims += 1; ok = False
        for cit in c["citations"]:
            cited += 1; s = all(supports(cit, n["value"]) for n in c["numbers"]); supported += s; ok |= s
        claims_with_support += ok
print(f"citation precision {supported}/{cited} | citation recall {claims_with_support}/{n_claims}")
```

```python
import csv
from sklearn.metrics import cohen_kappa_score
items = list(csv.DictReader(open("data/eval/judge_items.csv")))
judge = {l.item_id: l.supported for l in llm.structured(schemas.JudgeLabels, prompts.judge_messages([{"item_id": r["item_id"], "claim": r["claim"], "cited_text": r["cited_text"]} for r in items]), llm.P["llm"]["models"]["judge"]).labels}
a = [int(r["human_a"]) for r in items]; b = [int(r["human_b"]) for r in items]; j = [int(judge[r["item_id"]]) for r in items]
print(f"humans agree: kappa {cohen_kappa_score(a, b):.2f} | judge vs human A: kappa {cohen_kappa_score(a, j):.2f}")
```

**Where the humans disagree:** there are two items, not one: j16 and j20. The other 18 they agree on.

- **j16, "Revenue grew about 25% in FY2025" (A = 1, B = 0, judge = 1): rubric ambiguity.** The cited row (33.07 → 41.2) gives 24.6% growth, so "about 25%" is a correct rounding of a *derived* number. A accepted rounding; B required the exact figure. Neither is simply wrong: the rubric never says whether a rounded derived number counts as supported. Fix: write the convention down (for example, a derived number counts as supported if it is within the linter's tolerance after stated rounding) and re-label.
- **j20, "The contract term is 24 months" against "Term: 36 months" (A = 0, B = 1, judge = 1): an error.** 24 is not 36, so the claim is unsupported. A is right; **B and the judge are both wrong**. The judge's high κ against A (0.89) hides this mistake, and it matters, because a wrong contract term is exactly the kind of number Rachel will not accept.

**Score the judge against A, B, or their consensus?** Against the **consensus**: the 18 items both humans agree on. There the judge scores κ = 1.0 (18/18). Picking A or B alone means trusting one annotator's quirks, A's leniency on rounding or B's error on j20. The disagreement items should not be thrown away: resolve them by fixing the rubric (j16) or adjudicating (j20), then add them back. j20 should be reported as a known judge error, whatever κ says.

**Open question: how many human labels would convince Rachel to trust the judge on 500 memos?** Treat it as estimating the judge's agreement rate with a confidence interval. If the judge agrees with the human consensus on every one of n labels, the Wilson lower bound is n / (n + 1.96²). It takes **35** all-agree labels to put the lower bound at 90% agreement, and **73** to put it at 95%; any disagreements push the number higher. Our 18-item consensus (lower bound about 0.82) is not enough. The labels should also be a stratified random sample across memo sections and rubric lines, labelled by two people independently, since j20 shows the judge can be wrong in a way an aggregate κ hides.
